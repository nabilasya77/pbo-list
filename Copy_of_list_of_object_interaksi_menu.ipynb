```python
class Karyawan:
    def __init__(self, nik, nama, jabatan):
        self.nik = nik
        self.nama = nama
        self.jabatan = jabatan

    def __str__(self):
        return f"| {self.nik:10} | {self.nama:20} | {self.jabatan:15} |"


class SistemKaryawan:
    def __init__(self):
        self.database_karyawan = []

    def tambah_karyawan(self, karyawan_baru):
        if any(k.nik == karyawan_baru.nik for k in self.database_karyawan):
            return False, "NIK sudah terdaftar!"
        self.database_karyawan.append(karyawan_baru)
        return True, "Berhasil ditambahkan."

    def cari_karyawan(self, nik_cari):
        for k in self.database_karyawan:
            if k.nik == nik_cari:
                return k
        return None

    def tampilkan_semua(self):
        if not self.database_karyawan:
            print("\n[Data Karyawan Kosong]")
            return

        print("\n" + "="*60)
        print(f"| {'NIK':10} | {'NAMA':20} | {'JABATAN':15} |")
        print("-"*60)
        for k in self.database_karyawan:
            print(k)
        print("="*60)


def main():
    app = SistemKaryawan()

    while True:
        print("\n>>> SISTEM MANAJEMEN KARYAWAN <<<")
        print("1. Tambah Karyawan")
        print("2. Lihat Semua Karyawan")
        print("3. Cari Karyawan (NIK)")
        print("4. Keluar")

        pilihan = input("Pilih menu (1-4): ")

        if pilihan == "1":
            print("\n--- Input Data Karyawan ---")
            nik = input("Masukkan NIK: ")
            nama = input("Masukkan Nama: ")
            jabatan = input("Masukkan Jabatan: ")

            status, pesan = app.tambah_karyawan(Karyawan(nik, nama, jabatan))
            print(pesan)

        elif pilihan == "2":
            app.tampilkan_semua()

        elif pilihan == "3":
            nik_cari = input("\nMasukkan NIK yang dicari: ")
            hasil = app.cari_karyawan(nik_cari)
            if hasil:
                print(f"Ditemukan: {hasil.nama} - {hasil.jabatan}")
            else:
                print("Karyawan tidak ditemukan.")

        elif pilihan == "4":
            print("Program selesai. Terima kasih!")
            break

        else:
            print("Pilihan tidak valid!")


if __name__ == "__main__":
    main()
```
